Finding people's ADHD score

```python
import pandas as pd

data = pd.read_csv("Pop Music Survey (Responses) - Form Responses 1.csv")

x = ["I make careless mistakes at work or on homework.", "I have trouble staying focused on tasks.", "I forget daily responsibilities or assignments.", "I lose things like my keys and my phone.", "I get easily distracted by my surroundings.", "I continuously fidget when having to stay still.", "I interrupt people during conversations.", "I have a hard time waiting for my turn.", "I act without thinking things through.", "I talk excessively when it’s not appropriate", "I get emotional easily.", "I easily feel overwhelmed by tasks.", "I forget what I’m doing in the middle of tasks.", "I find it difficult to manage and prioritize my tasks.", "I struggle to break down a task into smaller steps."]

cleaned_adhd_data = data[x]

#adhd score
cleaned_adhd_data.astype(int)
cleaned_adhd_data.sum(axis = 1)
```

Finding music with complexity <= 3 and focusedness >= 4

```python
for idx, row in data.iterrows():
    songs = []
    
    for i in range(10):
        if i == 0:
            complexity = "How complex/chaotic was the music? (Scale from 1 to 5)"
            focus = "How focused did you feel while listening to the music? (Scale from 1 to 5)"
            predictable = "How predictable was the song's structure? (Scale from 1 to 5)"
        else:
            complexity = f"How complex/chaotic was the music? (Scale from 1 to 5).{i}"
            focus = f"How focused did you feel while listening to the music? (Scale from 1 to 5).{i}"
            predictable = f"How predictable was the song's structure? (Scale from 1 to 5).{i}"
        
        if complexity in data.columns and focus in data.columns:
            complexity = row[complexity]
            focus = row[focus]
            predictable = row[predictable]
            
            if complexity <= 2 and focus >= 4 and predictable >= 3:
                songs.append(i+1)
    
    if songs:
        name = row.get("What's your full name?", f"Responder {idx}")
        print(f"{name}: {songs}")
```
